```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Select categorical columns

categorical_cols = ['brand', 'country', 'model']
    
# Initialize One-Hot Encoder

encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Transform categorical variables

encoded_array = encoder.fit_transform(mobile_db[categorical_cols])

# Convert encoded data into a DataFrame

encoded_df = pd.DataFrame(
encoded_array,
columns=encoder.get_feature_names_out(categorical_cols),
index=mobile_db.index
)

# Display results

print("Original shape:", mobile_db.shape)
print("Encoded data shape:", encoded_df.shape)

print("\nEncoded columns:")
print(encoded_df.head())
```

```python
# Select numerical features from the original DataFrame
features = [
    'price_usd',
    'rating',
    'battery_life_rating',
    'camera_rating',
    'performance_rating',
    'design_rating',
    'display_rating'
]

X_numeric = mobile_db[features].copy()

# Combine numerical and one-hot-encoded categorical features
X = pd.concat([X_numeric, encoded_df], axis=1)

print("Combined feature shape:", X.shape)
print(X.head())

```

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Scale the selected features
X_scaled = scaler.fit_transform(X)

```

```python
from sklearn.cluster import KMeans

# Create the K-Means model
kmeans = KMeans(n_clusters=7, random_state=42, n_init=10)

# Fit the model and assign cluster labels
clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to the original dataset
mobile_db['cluster'] = clusters

# Check the results
print(mobile_db['cluster'].value_counts().sort_index())
```

```python
# Calculate average feature values for each cluster
cluster_analysis = mobile_db.groupby('cluster')[
    [
        'price_usd',
        'rating',
        'battery_life_rating',
        'camera_rating',
        'performance_rating',
        'design_rating',
        'display_rating'
    ]
].mean().round(2)

print(cluster_analysis)
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, mobile_db['cluster'])

print("Silhouette Score:", round(score, 4))
```

```python
import pandas as pd
import plotly.express as px
from sklearn.cluster import KMeans

k_values = range(2, 11)
inertia_values = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    inertia_values.append(model.inertia_)

elbow_df = pd.DataFrame({
    'Number of Clusters (K)': list(k_values),
    'WCSS': inertia_values
})

fig = px.line(
    elbow_df,
    x='Number of Clusters (K)',
    y='WCSS',
    markers=True,
    title='Elbow Method for Optimal Number of Clusters'
)

fig.update_layout(
    xaxis=dict(tickmode='linear', dtick=1)
)

fig.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 6):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=5000,
        random_state=42
    )

    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
from sklearn.preprocessing import StandardScaler

recommendation_features = [
    'price_usd',
    'battery_life_rating',
    'camera_rating',
    'performance_rating',
    'design_rating',
    'display_rating'
]

X_recommend = mobile_db[recommendation_features].copy()

scaler_recommend = StandardScaler()
X_recommend_scaled = scaler_recommend.fit_transform(X_recommend)

print("Recommendation data shape:", X_recommend_scaled.shape)
```

```python

from sklearn.metrics.pairwise import cosine_similarity

def recommend_mobile(model_name, top_n=10):

    # Find the index of the selected mobile model
    selected_index = mobile_db[
        mobile_db['model'] == model_name
    ].index[0]

    # Convert the DataFrame index to its row position
    selected_position = mobile_db.index.get_loc(selected_index)

    # Calculate cosine similarity
    similarity_scores = cosine_similarity(
        X_recommend_scaled[selected_position].reshape(1, -1),
        X_recommend_scaled
    ).flatten()

    # Exclude the selected record
    similarity_scores[selected_position] = -1

    # Get the most similar records
    top_positions = similarity_scores.argsort()[::-1][:top_n]

    # Display the selected mobile
    print("Selected mobile:")
    print(
        mobile_db.iloc[selected_position][
            ['brand', 'model', 'price_usd']
        ]
    )

    
    print("\nRecommended mobiles:")

    recommendations = mobile_db.iloc[top_positions][
        ['brand', 'model', 'price_usd', 'rating']
    ]

    # Remove duplicate mobile models
    recommendations = recommendations.drop_duplicates(
        subset=['brand', 'model']
    )

    print(recommendations)

```
